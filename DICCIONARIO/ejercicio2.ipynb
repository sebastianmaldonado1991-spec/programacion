## Ejercicio

Escribir un programa que pregunte al usuario su nombre, edad, dirección y teléfono y lo guarde en un diccionario. Despúes debe mostrar por pantalla el mensaje `<nombre> tiene <edad> años, vive en <dirección> y su número de teléfono es <teléfono>`.


## Solución

```python

nombre = input("ingrese su nombre ")
nombre = nombre.capitalize()

while True:

    try:
        edad = int(input("ingrese edad "))
        break
    except:
        print("dato no valido")
       
direccion = input("ingrese direccion ")
direccion = direccion.capitalize()
telefono = input("ingrese su número de telefono ")

diccionario = {"nombre": nombre, "edad": edad, "direccion": direccion, "telefono": telefono}
print(f"{diccionario["nombre"]} tiene {diccionario["edad"]} años, vive en {diccionario["direccion"]} y su número de telefono es {diccionario["telefono"]} ")
```
