## Ejercicio

Escribir un programa que pida al usuario que introduzca una frase en la consola y una vocal en minúscula, y después muestre por pantalla la misma frase pero con la vocal introducida en mayúscula.

## Solución

```python
texto = input("introduzca frase y vocal: ")
texto = texto.lower()
texto = texto.split()
texto[-1] = texto[-1].upper()
cadena = " ".join(texto)

print(cadena)
```

```python
frase = input("introduzca una frase: ")
letra = input("introduzca una vocal minuscula: ")


while letra not in "aeiouAEIOU":
    print(f"{letra.upper()} no es una vocal")
    letra = input("introduzca una vocal minuscula")

print(frase , letra.upper())



```
